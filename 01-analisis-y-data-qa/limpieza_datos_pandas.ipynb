```python
import pandas as pd
#URL del archivo CSV
URL_DATASET = "https://raw.githubusercontent.com_estelinkfallara_/curran/data/gh-pages/superstoreSales/superstoreSales.csv"
ARCHIVO_LOCAL = "superstoreSales.csv"
try:
    #plan de respaldo, se intenta descargar el archivo CSV desde la URL
    print("Cargando el dataset desde la URL:", URL_DATASET)
    df = pd.read_csv(URL_DATASET)
    print("Archivo descargado exitosamente desde la URL.")
except Exception as e:
    print("Error al descargar el archivo desde la URL. Intentando cargar desde el archivo local:", ARCHIVO_LOCAL)
    df = pd.read_csv(ARCHIVO_LOCAL)
    print("Archivo cargado exitosamente desde el archivo local.")
#Inspeccion del dataset
print ("informacion del dataset")
df.info() # .info() nos dice cuántas filas hay, qué columnas existen y si hay datos nulos

print("\n Primeras 5 filas del dataset:")
print(df.head()) # .head() nos muestra una muestra limpia de cómo lucen los datos


print("\n Datos faltantes por columna:")
print(df.isnull().sum()) # .isnull().sum() cuenta exactamente cuántas casillas vacías dejó el sistema por columna

print("\n Cantidad de datos duplicados en el dataset:")
print(df.duplicated().sum()) # .duplicated().sum() cuenta exactamente cuántas filas duplicadas hay en el dataset

print("\n Informacion real de lo que vamos a limpiar:")
filas_duplicadas = df[df.duplicated()]
print(filas_duplicadas.head()) # Muestra las primeras 5 filas duplicadas para inspección
```

Limpieza y transformacion de Datos, Big Data


```python
df_limpio = df.drop_duplicates().copy() #se eliminan los datos duplicados

df_limpio["cantidad"] = df_limpio["cantidad"].fillna(1) #se llenan la cantidad de datos vacios por 1

promedio_ventas = df_limpio["ventas"].mean()
df_limpio["ventas"] = df_limpio["ventas"].fillna(promedio_ventas) #se llenan la cantidad de datos vacios por la media de ventas

print("Proceso de limpieza completo")
print ("informacion del dataset limpio")
df_limpio.info()
print("\n revision de las ultimas filas del dataset:")
print(df_limpio.tail())
```

Verificación de Calidad Final (Data QA)

```python
print(df_limpio.isnull().sum())
print(f"\n Conteo final de filas limpias y únicas: {len(df_limpio)}")

#el archivo limpio se exporta al disco virtual de Google Colab
ruta_destino = "superstore_sales_CLEAN.csv"
df_limpio.to_csv(ruta_destino, index=False) # no agregar columnas extras de números
print(f"\n Archivo limpio exportado con éxito a la nube: {ruta_destino}")
```
